```python
# Probability of finishing reports on time using Monte Carlo Simulations
```

There is a meeting happening in **_8 hours_** from now but i have two reports to submit.  
From previous reports I have written, it will take me 1-5 hours to write the first report and 2-4 hours to write the second one.  
With this information, I want to find out the my chance of finishing the two reports on time and making it to the meeting 

```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

I will run **_100,000 simulations_** to find out the probility on finishing the reports before the meeting in 7hours

```python
sim = 100000
meeting_time = 8
```

Time to generate random times each report is likely to take  
Since the time to write each report is not dependant on the other,  
I will use a **_uniform distribution_**

```python
reportA = np.random.uniform(1,5,sim)
reportB = np.random.uniform(2,6,sim)

duration = reportA + reportB   #total time to complete both reports
```

```python
plt.hist(duration,density=True)
plt.axvline(meeting_time, color='red', linestyle='dashed')

plt.xlabel("Total Report Time (hours)")
plt.ylabel("Frequency")
plt.title("Monte Carlo Simulation of Report Completion Time")

plt.show()
```

From the histogram, it is clear that the **_average time in completing both reports is 7 hours_**.  
The extreme ends of showing completion times at 3 and 11 hours but with very low frequency.  
The red dotted line represents the hours before my meeting which is greater than the mean completion time.  
This tells me i my chances of finishing the reports before time are high.

```python
probability_of_success = ((duration < 8).sum()/sim)  #probability that I finish both reports before my meeting
print(probability_of_success)
```

Approximately **72%** is the probability of finishing both reports before my meeting.
